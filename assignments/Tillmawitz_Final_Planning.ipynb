For the final project I intend to use one of the [Amazon Reviews](https://amazon-reviews-2023.github.io/main.html) datasets, likely the Software category as it has ~5 million ratings in the Pure IDs dataset as well as 1.3 million ratings in the 5-Core dataset. The 5-Core datasets consist of only the users and items with a minimum of 5 reviews, whereas the Pure IDs version of the dataset consists of all users and items regardless of number of reviews. There are pros and cons to each dataset in terms of distribution, diversity, noise, and quality which could be interesting to explore if both datasets are able to be processed as part of the project, but if the Pure IDs data is too much then I will still be able to use the 5-Core data to complete the project. Should I choose to convert ratings to implicit values I will likely instead use a category such as Books which has ~30 million Pure ID and ~9.5 million 5-core reviews which, once converted to implicit values indicating high ratings, should still be large enough to meet the requirements of the project.

To train the model I will be using [PyTorch](https://pytorch.org/blog/introducing-torchrec/) as well as [Google Colab](https://colab.google/). Google Colab is a Jupyter Notebook hosting service that can be easily integrated with GCP resources, which should make accessing GPUs and utilizing a distributed cluster easy similar to working with Databricks. It is possible to use Google Drive to host the data for the project which is a reasonable stand in for an S3 bucket as the method of accessing the data is very similar on both platforms. PyTorch is a newer machine learning library focussed on deep learning and neural networks and is designed to make use of distributed platforms and GPUs. It is quickly gaining popularity for training models at a large scale, and given my professional background involved extensive work with Spark I am opting for a distributed platform I have less familiarity with. Depending on how mature the Google Colab to GCP deployment pipeline is I may deploy the model to GCP or to AWS in order to meet the requirements for Project 6 as well.

In terms of model development, I plan on iteratively developing models and featuresets until resource bottlenecks grow too large to continue. To begin with I will use matrix factorization to generate features based on ratings and training a matrix factorization based model as a baseline. I will then add neural network layers to use Neural Collaborative Filtering to attempt to better model non-linear relationships in the data. Performance permitting I intend to move on to generating embeddings using autoencoders or more advanced fully integrated models such as a two-tower model. Experimenting with feature generation by processing metadata such as review text, feature descriptions, price, and other information might be included depending on model performance and available resources, however this is a stretch goal.
